```python
import warnings
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
from google.colab import drive
import seaborn as sns
import plotly.express as px
```

READING DATA FILES

```python
file1=pd.read_csv("/content/drive/MyDrive/mental-and-substance-use-as-share-of-disease.csv")
file2=pd.read_csv("/content/drive/MyDrive/prevalence-by-mental-and-substance-use-disorder.csv")
```

```python
file1.head()
```

```python
file2.head()
```

MERGING FILE1 AND FILE2

```python
data=pd.merge(file1,file2)
data.head(20)
```

REMOVING NON NUMERIC COLOUMNS

```python
data.isnull().sum()
```

```python
data.drop('Code',axis=1,inplace=True)
data.head()

```

SIZE OF DATA

```python
print(data.shape)
print(data.size)
```

HEATMAP OF GIIVEN DATA

```python
sns.heatmap(data.corr(),annot=True,cmap='Greens')
```

```python
data.info()
```

CHANGING NON NUMERIC LABELS TO NUMERRIC LABELS


```python
from sklearn.preprocessing import LabelEncoder
l=LabelEncoder()
for i in data.columns:
  if data[i].dtype == 'object':
    data[i]=l.fit_transform(data[i])
```

```python
data.head()
```

SPLITTING DATA INTO TEST AND TRAIN

```python
x=data.drop('DALYs (Disability-Adjusted Life Years) - Mental disorders - Sex: Both - Age: All Ages (Percent)',axis=1)
y=data['DALYs (Disability-Adjusted Life Years) - Mental disorders - Sex: Both - Age: All Ages (Percent)']

from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=20,random_state=2)
```

```python
print("xtrain:",xtrain.shape)
print("xtest:",xtest.shape)
print("ytrain:",ytrain.shape)
print("ytest:",ytest.shape)
```

USING LINEAR REGRESSION

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error,r2_score
lr=LinearRegression()
lr.fit(xtrain,ytrain)

ytrain_pred=lr.predict(xtrain)

mse=mean_squared_error(ytrain,ytrain)

rmse=(np.sqrt(mean_squared_error(ytrain,ytrain_pred)))

r2=r2_score(ytrain,ytrain_pred)

print("------------------------------------------")
print("LINEAR REGRESSION PERFORMANCE")
print('mse is',mse)
print('rmse is',rmse)
print('r2 score is',r2)
```

USING RANDOM FOREST REGRESSOR

```python
from sklearn.ensemble import RandomForestRegressor
rf=RandomForestRegressor()
rf.fit(xtrain,ytrain)

ytrain_pred=rf.predict(xtrain)
mse=mean_squared_error(ytrain,ytrain_pred)
rmse=(np.sqrt(mean_squared_error(ytrain,ytrain_pred)))
r2=r2_score(ytrain,ytrain_pred)

print("--------------------------------")
print("RANDOM FOREST REGRESSOR PERFORMANCE")
print('mse is',mse)
print('rmse is',rmse)
print("r2 score is",r2)
```

EVALUATING TEST AND TRAIN MODELS

```python
ytest_pred=lr.predict(xtest)
mse=mean_squared_error(ytest,ytest_pred)
rmse=(np.sqrt(mean_squared_error(ytest,ytest_pred)))
r2=r2_score(ytest,ytest_pred)

print("linear regression performance for testing set")
print('mse is',mse)
print('rmse is',rmse)
print('r2 score is',r2)

print()
print()
print()
print()


ytest_pred=rf.predict(xtest)
mse=mean_squared_error(ytest,ytest_pred)
rmse=(np.sqrt(mean_squared_error(ytest,ytest_pred)))
r2=r2_score(ytest,ytest_pred)

print("random forest regressor performance for testing set")
print('mse is',mse)
print('rmse is',rmse)
print('r2 score is',r2)

```
